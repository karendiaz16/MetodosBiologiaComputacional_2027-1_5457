```python
#Una cepa de "E.coli" iene una tasa de crecimieno tal que unindividuo se duplica cada 30 min
#Calcula  en cuanto tiempo esta cepa:
#a) Llenaria el volumen de la Tierra con su radio (considerandola como una esfera hueca) 
#b) cubrirala superficie de la Tierra
#c) repite el calculo para una tasa de crecimiento
  #a) se duplica cada hora
  #b) se duplica cada media hora
```

```python
import math as mt
#El radio de la Tierra en km
rT= (6371e3)**3
#El radiode E.coli en um
rE= (0.5e-6)**3
#Calculo para saber cuantos caben por despeje de volumen nos queda:
N= rT / rE
print (N)
```

```python
#Tiempo en 1/2 de hora
t=mt.log2(N)
print (t)
```

```python
#Tiempo en 1 hora
Th=(t/2)
print (Th)
#Tiempo en dias
Td=(Th/24)
print (Td)
```

```python
#Superficie de la Tierra (basada en el radio al cuadrado en metros)
sT= (6371e3)**2
#Superficie ocupada por una E.coli 
sE= (0.5e-6)**2
#Calculo para saber cuantas bacterias senecesitan para cubrir la superficie de la Tierra
Nb = sT/sE
#Tiempo en  1/2 horas
Ts= mt.log2 (Nb)
print (Nb)
print (Ts)
```

```python
#Tiempo x horas
Tsh= (Ts/2)
print (Tsh)
#Tiempo en dias
Tsd=(Tsh/24)
print (Tsd)
```
