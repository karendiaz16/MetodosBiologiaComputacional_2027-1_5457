### **Ejercicio:** Logístico hasta aproximarse a K

Simula $N_{t+1} = N_t + r N_t(1-N_t/K)$ con $N_0 = 100$, $r = 0.3$ y $K = 1000$. Detén el proceso en la primera generación con $N \leq  0.95 K$, o al alcanzar `1000` iteraciones. Guarda los tamaños en una lista. ¿Se aproxima la población a $K$ sin alcanzarla exactamente? 

````python
N0 = 100
r = 0.3
K = 1000

N = N0
generaciones = 0
historial = [N]

while N < 0.95 * K and generaciones < 1000:
    N = N + r * N * (1 - N / K)
    generaciones = generaciones + 1
    historial.append(N)

print("Generaciones necesarias:", generaciones)
print("Tamano final de la poblacion:", N)
````

*Hint:* La forma más sencilla es usar el teorema de Pitagoras.

````python
import math

def distancia_cilindrica(r, theta, z):
    x = r * math.cos(theta)
    y = r * math.sin(theta)
    d = math.sqrt(x**2 + y**2 + z**2)
    return d

print(distancia_cilindrica(3, 0, 4))
print(distancia_cilindrica(5, 1.2, 0))
print(distancia_cilindrica(2, 3.1416, 7))
````

<h4 align="center">$\displaystyle n! = \prod_{i=1}^n i$</h4> 

Y pruebala para diferentes valores.

````python
def factorial(n):
    resultado = 1
    for i in range(1, n + 1):
        resultado = resultado * i
    return resultado

print(factorial(5))
print(factorial(0))
print(factorial(7))
````

#### **_Recursividad._**

Una característica muy útil de las funciones que podemos definir es la **_recursividad_**, es decir, la capacidad de una función de llamarse a sí misma. Por ejemplo, considera la siguiente definición del factorial $n!$:

<h4 align="center">
$
      n! = \biggl\lbrace\begin{array}{ll}
        1 & \qquad\mbox{si $n=1$,} \\
        n\times(n-1)! & \qquad\mbox{si $n>1$.}
      \end{array}
$
</h4> 
      
Esto constituye una definición completa del factorial que nos permite calcular el valor de $n!$ para cualquier entero positivo. 


### **Ejercicio:** 

**a)** A partir del ejercicio anterior, convierte (si es que no lo hiciste ya) tu función en una _función recursiva_, es decir, que  cuando la función tome como argumento un número $n$ diferente de $1$, la función se llame a sí misma para calcular el factorial de $n-1$.

````python
def factorial_recursivo(n):
    if n == 1:
        return 1
    else:
        return n * factorial_recursivo(n - 1)

print(factorial_recursivo(5))
print(factorial_recursivo(7))
````

**b)** Escribe una función, **usando recursividad**, que calcule la suceción de Fibonacci.

````python
def fibonacci(n):
    if n == 1 or n == 2:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

for i in range(1, 11):
    print(fibonacci(i))
````

**c)** Euclides demostró que el máximo común divisor $g(m,n)$ de dos enteros no negativos $m$ y $n$ satisface que:

<h4 align="center">
$
    g(m,n) = \biggl\lbrace\begin{array}{ll}
      m & \qquad\mbox{si $n=0$,} \\
      g(n,m\>\textrm{mod}\>n) & \qquad\mbox{si $n>0$.}
    \end{array}
$
</h4>

Usando esta fórmula escribe una función `g(m,n)` que emplee **recursividad** para calcular el máximo común divisor de $m$ y $n$. Usa tu función para calcular e imprimir el máximo común divisor de 108 y 192.

````python
def g(m, n):
    if n == 0:
        return m
    else:
        return g(n, m % n)

print("El maximo comun divisor de 108 y 192 es:", g(108, 192))
````

En la mayoría de los casos, si una cantidad se puede calcular **_sin_ recursividad**, será más rápido y normalmente se recomienda hacerlo así, si es posible. Sin embargo, hay algunos cálculos que son esencialmente imposibles (o al menos mucho más difíciles) sin recursividad.

Supongamos que tenemos un número entero $n$ y queremos saber sus **_factores primos_**. Los factores primos se pueden calcular con relativa facilidad dividiendo repetidamente por todos los números enteros desde 2 hasta $n$ y verificando si el resto es cero. 

**a)** Escribe una función que tome como argumento un número $n$ y devuelva una lista de sus factores primos.

````python
def factores_primos(n):
    factores = []
    d = 2
    while d <= n:
        if n % d == 0:
            factores.append(d)
            n = n // d
        else:
            d = d + 1
    return factores

print(factores_primos(60))
print(factores_primos(97))
print(factores_primos(360))
````

**b)** Usando la función anterior, escribe un programa que imprima los números primos hasta cualquier límite que elijas verificando si tienen un solo factor primo. 

````python
N = 50

for numero in range(2, N + 1):
    factores = factores_primos(numero)
    if len(factores) == 1:
        print(numero)
````
